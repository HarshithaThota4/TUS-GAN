```python
import pandas as pd

# Replace 'your_file.txt' with the actual filename
df = pd.read_csv("E:\TUS-raw(2019)\TUS106_L01.TXT", header=None, usecols=[0])  # Read only first column

# Extract first 32 alphabets
df['Schedule_Id'] = df[0].str[:3]
df['FSU_Serial_No'] = df[0].str[3:8]
df['Schedule'] = df[0].str[8:11]
df['Survey_Year'] = df[0].str[11:15]
df['Sector'] = df[0].str[15]
df['NSS_Region'] = df[0].str[16:19]
df['District'] = df[0].str[19:21]
df['Stratum'] = df[0].str[21:23]
df['Sub_Stratum'] = df[0].str[23:25]
df['Sub_Round'] = df[0].str[25]
df['FOD_Sub_Region'] = df[0].str[26:30]
df['Sample hhld. No.'] = df[0].str[30:32]
df['Common_Id'] = df[0].str[:32]
df['Level'] = df[0].str[32:34]
df['Filler'] = df[0].str[34:39]
df['Informant Sl.No.'] = df[0].str[39:41]
df['Response Code'] = df[0].str[41]
df['Survey Code'] = df[0].str[42]
df['Substitution Code/ Casualty code'] = df[0].str[43]
df['Filler'] = df[0].str[44:56]
df['Date of Survey'] = df[0].str[56:62]
df['Date of Despatch'] = df[0].str[62:68]
df['Time to canvass(minutes)'] = df[0].str[68:71]
df['No. of investigators(FI/ASO) in the team'] = df[0].str[71]
df['Remarks in block 7'] = df[0].str[72]
df['Remarks in block 8'] = df[0].str[73]
df['Remarks elsewhere in Sch.'] = df[0].str[74]
df['Remarks elsewhere in Sch.'] = df[0].str[75]
df['Blank'] = df[0].str[76:126]
df['NSC'] = df[0].str[126:129]
df['MULT'] = df[0].str[129:139]





# Save to a CSV file (drop original data column)
df.drop(columns=[0], inplace=True)  # Drop the first column (original data)
df.to_csv('File1_updated.csv', index=False)
```

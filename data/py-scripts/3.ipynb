```python
import pandas as pd

# Replace 'your_file.txt' with the actual filename
df = pd.read_csv('/Users/vjsuyash/Documents/UGP/Data Preprocessing/File3/TUS106_L03 copy.TXT', header=None, usecols=[0])  # Read only first column

# Extract first 32 alphabets
df['Common_Id'] = df[0].str[:32]
df['Level'] = df[0].str[32:34]
df['Filler'] = df[0].str[34:39]
df['Household size'] = df[0].str[39:42]
df['religion'] = df[0].str[42]
df['Social group '] = df[0].str[43]
df['Land possessed as on date of survey(code)'] = df[0].str[44:46]
df['usual consumer expenditure in a month for household purposes out of purchase (A)'] = df[0].str[46:56]
df['imputed value of usual consumption in a month from home grown stock (B)'] = df[0].str[56:66]
df['imputed value of usual consumption in a month from wages in kind, free collection, gifts, etc (C )'] = df[0].str[66:76]
df['expenditure on purchase of household durable during last 365 days (D)'] = df[0].str[76:86]
df['usual monthly consumer expenditure E: [A+B+C+(D/12)]'] = df[0].str[86:96]
df['Primary source of energey for cooking'] = df[0].str[96:98]
df['Primary source of energey for lighting'] = df[0].str[98]
df['Type of washing of clothes'] = df[0].str[99]
df['Type of sweeping of floor'] = df[0].str[100]
df['Type of structure of the dwelling unit'] = df[0].str[101]
df['member of age 5 years and above needing special care but no care giver is available'] = df[0].str[102]
df['Blank'] = df[0].str[103:126]
df['NSC'] = df[0].str[126:129]
df['MULT'] = df[0].str[129:139]



# Save to a CSV file (drop original data column)
df.drop(columns=[0], inplace=True)  # Drop the first column (original data)
df.to_csv('File3.csv', index=False)
```

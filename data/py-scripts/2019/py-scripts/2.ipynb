import pandas as pd

# Replace 'your_file.txt' with the actual filename
df = pd.read_csv('/Users/vjsuyash/Documents/UGP/Data Preprocessing/File2/TUS106_L02 copy.TXT', header=None, usecols=[0])  # Read only first column

# Extract first 32 alphabets
df['Common_Id'] = df[0].str[:32]
df['Level'] = df[0].str[32:34]
df['Filler'] = df[0].str[34:36]
df['Person_Serial_No'] = df[0].str[36:39]
df['Relation_to_head'] = df[0].str[39]
df['Gender'] = df[0].str[40]
df['Age'] = df[0].str[41:44]
df['Marital_Status'] = df[0].str[44]
df['Highest_level_of_education'] = df[0].str[45:47]
df['usual principal activity: status (code)'] = df[0].str[47:49]
df['industry of work: 2-digit of NIC 2008'] = df[0].str[49:51]
df['Blank'] = df[0].str[51:126]
df['NSC'] = df[0].str[126:129]
df['MULT'] = df[0].str[129:139]



# Save to a CSV file (drop original data column)
df.drop(columns=[0], inplace=True)  # Drop the first column (original data)
df.to_csv('extracted_data.csv', index=False)

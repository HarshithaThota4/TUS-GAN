```python
import pandas as pd

# Replace 'your_file.txt' with the actual filename
df = pd.read_csv('TUS106_L05 copy.txt', header=None, usecols=[0])  # Read only first column

# Extract first 32 alphabets
df['Common_Id'] = df[0].str[:32]
df['Level'] = df[0].str[32:34]
df['Filler'] = df[0].str[34:36]
df['Serial no.of member'] = df[0].str[36:39]
df['age'] = df[0].str[39:42]
df['srl. No of activity'] = df[0].str[42:45]
df['time from'] = df[0].str[45:50]
df['time to'] = df[0].str[50:55]
df['whether performed multiple activity in the time slot'] = df[0].str[55]
df['whether simultaneous activity'] = df[0].str[56]
df['whether a major activity'] = df[0].str[57]
df['3-digit activity code'] = df[0].str[58:61]
df['where the activity was performed'] = df[0].str[61]
df['unpaid/paid status of activity'] = df[0].str[62:64]
df['enterprise type'] = df[0].str[64]
df['Blank'] = df[0].str[65:126]
df['NSC'] = df[0].str[126:130]
df['MULT'] = df[0].str[130:139]



# Save to a CSV file (drop original data column)
df.drop(columns=[0], inplace=True)  # Drop the first column (original data)
df.to_csv('File5.csv', index=False)
```

```python
import pandas as pd
import numpy as np

# Load the data into a DataFrame
df = pd.read_csv('File5.csv')

# Replace non-finite values with a suitable value (e.g., 0)
df['Serial no.of member'] = df['Serial no.of member'].replace([np.inf, -np.inf], 0)
df['Serial no.of member'] = df['Serial no.of member'].fillna(0)

# Convert 'Serial no.of member' to integer
df['Serial no.of member'] = df['Serial no.of member'].astype(int)

# Create 'Person_ID' column
df['Person_ID'] = df['Common_Id'].astype(str) + '_' + df['Serial no.of member'].astype(str)


df.to_csv('File5.csv', index=False)
```

```python
import matplotlib.pyplot as plt

# Filter the DataFrame and calculate value counts
filtered_df = df[df['3-digit activity code'].isin([119, 129, 139, 149, 159, 169, 219, 229, 319, 329])]
value_counts = filtered_df['3-digit activity code'].value_counts()

# Create a bar plot
plt.figure(figsize=(10, 6))
value_counts.plot(kind='bar', color='skyblue')
plt.title('Value Counts of 3-Digit Activity Codes')
plt.xlabel('Activity Code')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
```

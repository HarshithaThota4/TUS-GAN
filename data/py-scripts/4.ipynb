import pandas as pd

# Replace 'your_file.txt' with the actual filename
df = pd.read_csv('TUS106_L04 copy.TXT', header=None, usecols=[0])  # Read only first column

# Extract first 32 alphabets
df['Common_Id'] = df[0].str[:32]
df['Level'] = df[0].str[32:34]
df['Filler'] = df[0].str[34:36]
df['srl. No of member'] = df[0].str[36:39]
df['Age'] = df[0].str[39:42]
df['relationship of the informant with the household member '] = df[0].str[42:44]
df['day of week'] = df[0].str[44]
df['type of the day'] = df[0].str[45]
df['response code'] = df[0].str[46]
df['Blank'] = df[0].str[47:126]
df['NSC'] = df[0].str[126:129]
df['MULT'] = df[0].str[129:139]



# Save to a CSV file (drop original data column)
df.drop(columns=[0], inplace=True)  # Drop the first column (original data)
df.to_csv('File4.csv', index=False)

```python
import pandas as pd

# Household level
f1 = pd.read_csv('File1.csv', usecols=['Common_Id', 'District'])
f2 = pd.read_csv('file2.csv', usecols=['Common_Id', 'Gender', 'Age', 'Marital_Status',
                                        'usual principal activity: status (code)',
                                        'industry of work: 2-digit of NIC 2008'])
f3 = pd.read_csv('File3.csv', usecols=['Common_Id', 'Household size', 'Social group ',
                                        'usual monthly consumer expenditure E: [A+B+C+(D/12)]',
                                        'Primary source of energey for cooking',
                                        'Primary source of energey for lighting',
                                        'Type of structure of the dwelling unit'])

# Member × Day level
f4 = pd.read_csv('File4.csv', usecols=['Common_Id', 'srl. No of member',
                                        'relationship of the informant with the household member ',
                                        'day of week', 'type of the day'])

# Activity level — merge with f4 on BOTH Common_Id + member serial
f5 = pd.read_csv('File5.csv', usecols=['Common_Id', 'Serial no.of member',
                                        'srl. No of activity', 'time from', 'time to',
                                        'whether performed multiple activity in the time slot',
                                        'whether simultaneous activity',
                                        'whether a major activity',
                                        '3-digit activity code',
                                        'where the activity was performed',
                                        'unpaid/paid status of activity', 'enterprise type'])

# Step 1: merge activity-level files first (f4 × f5 on member key)
f4_f5 = pd.merge(
    f4, f5,
    left_on=['Common_Id', 'srl. No of member'],
    right_on=['Common_Id', 'Serial no.of member'],
    how='inner'
).drop(columns=['Serial no.of member'])

# Step 2: bring in household-level info
merged = f1.merge(f2, on='Common_Id', how='inner') \
           .merge(f3, on='Common_Id', how='inner') \
           .merge(f4_f5, on='Common_Id', how='inner')

merged.to_csv('merged_file.csv', index=False)
```
